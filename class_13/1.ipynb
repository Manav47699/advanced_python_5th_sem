```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([10, 20, 30, 40])
y = np.array([14, 16, 17, 18])
```

```python
X = np.arange(1, 30, 0.2)
y = 1.123*np.square(X) + 2.456

plt.figure(figsize=(6, 4))
plt.plot(X, y, c='r')
plt.grid(True)
plt.show()
```

```python
#plots using subplots
# fig, ax = plt.subplots()
fig=plt.figure()
axis1 = fig.add_subplot(2, 2, 1)

axis1.plot(X, y, c='r')
axis1.set_xlabel("X")
axis1.set_ylabel("y")
axis1.set_title("X vs y")

axis2 = fig.add_subplot(2, 2, 2)

axis2.plot(X, y, c='r')
axis2.set_xlabel("X")
axis2.set_ylabel("y")
axis2.set_title("X vs y")

plt.tight_layout()
```

```python
fig, axes = plt.subplots(2, 2)
fruits = ["apples", "banana", "grapes", "guava"]
quantity = [1, 2, 3, 4]
for i in range(2):
    for j in range(2):
        axes[i, j].bar(fruits, quantity, color={"Red", "Yellow", "Green", "Blue"})
        axes[i, j].set_title(f"plot{i}, {j}")
print(axes)
print(fig)
plt.tight_layout()
plt.legend()
# fig.subplots_adjust(wspace=0, hspace=3)   #yesle chai manually spacing adjust garnu milxa
```

# histogram

```python
data = np.random.normal(loc=50, scale=10, size=1000)

plt.hist(data, bins=20, color='skyblue', edgecolor='black')

plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of Random Data')
plt.show()
```

```python
from sklearn.datasets import make_moons

X, y = make_moons(
    n_samples=1000,
    noise=0.03,
    random_state=42,
)

# print(X[:3], y[:3])

fig, axes = plt.subplots()
axes.scatter(X[:,0], X[:,1], c=y)
# axes.text(x=0.6, y=1, "Hellow world", family = "monospace", fontsize= 10)
print(X.shape)
print(y.shape)
axes.set_title("scatterplot")

```

```python
from sklearn.datasets import load_iris
data = load_iris(as_frame=True)
X = data.data
y = data.target
feature_names = data.feature_names
class_names = data.target_names
```

```python
print (X[:2], y[:2])
print(feature_names)
print(class_names)
```

```python
fig=plt.figure()
axis1 = fig.add_subplot(2, 2, 1)
axis1.plot(X[:0], y[:0], c='r')
axis1.set_xlabel("X")
axis1.set_ylabel("y")
axis1.set_title("X vs y")

axis2 = fig.add_subplot(2, 2, 2)

axis2.plot(X, y, c='r')
axis2.set_xlabel("X")
axis2.set_ylabel("y")
axis2.set_title("X vs y")

plt.tight_layout()
```

```python
import matplotlib.pyplot as plt
```

```python
fig, axes = plt.subplots(2, 2)

X, y = load_iris(return_X_y= True, as_frame=True)

axes[0, 0].bar(x=['Setosa', 'Versicolor', 'Virginica'], 
               height=y.value_counts(), color=["Red", "Green", "Blue"])

axes[0,0].set_title("Species Distribution", fontsize=8)

axes[0, 1].scatter(X['sepal length (cm)'], X['petal length (cm)'], c=y)
axes[0, 1].set_title("Scatter plot sepal vs petal length", fontsize=8)

axes[1, 0].hist(X['sepal length (cm)'], bins=10, edgecolor='black')
axes[1, 0].set_title("histogram of sepal length", fontsize=8)

plt.tight_layout()
```
