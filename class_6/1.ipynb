```python
import requests
import pandas as pd

API_key = "YOUR_API_KEY"
city_names = ["Dharan", "Itahari", "Biratnagar", "Pokhara", "Kathmandu", "Damak", "Janakpur"]

weather_data_list = []

for city in city_names:
    url = f"https://api.openweathermap.org/data/2.5/weather?q={city}&appid={API_key}&units=metric"
    response = requests.get(url)
    
    if response.status_code == 200:
        data = response.json()
        # Flatten JSON and append dictionary to list
        weather_data_list.append({
            "name": data["name"],
            "temp": data["main"]["temp"],
            "humidity": data["main"]["humidity"],
            "description": data["weather"][0]["description"],
            "feels_like": data["main"]["feels_like"],
            "wind_speed": data["wind"]["speed"],
            "clouds": data["clouds"]["all"]
        })
    else:
        print(f"Failed to get data for {city}, status code: {response.status_code}")

# Convert list of dicts to DataFrame
df_weather = pd.DataFrame(weather_data_list)
print(df_weather)
```
