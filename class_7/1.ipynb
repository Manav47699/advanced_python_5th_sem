```python
import requests
from bs4 import BeautifulSoup

url = "https://sports.ndtv.com/cricket"
response = requests.get(url)

if response.status_code == 200:
    data = response.text   # <-- get the HTML content as a string
    soup = BeautifulSoup(data, "html.parser")
    print(soup.prettify())
else:
    print("Error:", response.status_code)


```

```python
print(soup.h2.text)
news = soup.find_all("h2")
for n in news:
    print (n)
```

```python
import pandas as pd
final_data =[]
for val in news:
    text_data = val.text
    link_data = val.a
    if link_data:
        link = link_data("href")
    if len(text_data) >20:
        final_data.append({
            "headline": text_data,
            "link":link_data
        })
df = pd.DataFrame(final_data)
df.to_csv("news.csv", index= False)
```
