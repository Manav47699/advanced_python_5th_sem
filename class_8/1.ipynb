```python
import pandas as pd

df_A = pd.DataFrame({
    'Temp': [20, 30, 40, 50]},
    index = ['Day1', "Day2", "day3", "day4"],
)

df_B = pd.DataFrame({
    'Humidity': [120, 130, 140, 150]},
    index = ['Day1', "Day2", "day5", "day7"],
)
```

```python
joined_df = df_A.join(df_B, how ='inner')
print (joined_df)
```

```python
joined_df = df_A.join(df_B, how ='outer')
print (joined_df)
```

```python
joined_df = df_A.join(df_B, how ='left')
print (joined_df)
```

```python
wide_df = pd.DataFrame({
    'maths': [1000, 2000, 3000, 4000],
    'science': [100, 200, 300, 400],
    'english': [10, 20, 30, 40],
    'studnets': ["Manav", "Mausam", "himal", "kshitiz"]
})
print(wide_df)
```

```python
#melt converts columns into rows
long_df = pd.melt(
    frame = wide_df, 
    id_vars=["studnets"],
    # value_vars=['science', 'math', 'english'],
    var_name="Subjects",
    value_name="Marks",
)

print(long_df)
```

```python
df_try_pivot = long_df.pivot(
    index="studnets",
    columns="Subjects",
    values="Marks",
)
print(df_try_pivot)
```

```python
df_try_stack = long_df.stack()
print (df_try_stack)
```

```python
df_try_unstack = long_df.unstack()
print (df_try_unstack)
```

```python
df_try_pivottable = long_df.pivot_table(
    # data="long_df",
    index="studnets",
    columns="Subjects",
    values="Marks",
    aggfunc="mean"
)
print (df_try_pivot)
```
