Handling real world values -> missing values, catagorical values etc

```python
import pandas as pd
```

```python
df = pd.read_csv("missing.csv")
print (df)
```

```python
print (df.isna().sum())
# value xa bhane false, xaina bhane true return hunxa
# sum() le kati ota value missing xa bhanera dekhauxa
```

```python
df.dropna()         #drops rows with missing values
df.head()
```

```python
print (df)
```

```python
df = df.dropna(axis=1)  # drops columns with missing values
df.head()
```

```python
print (df)
```

```python
df["age"] = df["age"].fillna(df["age"].mean())    #fill the missing value in age column with their mean
df["name"] = df["name"].fillna("Unknown")          # fill the missing values in name column with "unknown"
```

```python
df.to_csv('missing_value_fixed.csv')
```

# catagorical data


```python
df = pd.DataFrame({
    "EID": [101, 102, 103, 104],
    "Name": ["manav", "mausam", "himal", "ram"],
    "level": ["junior", "Obber juniro", "lastai junior","jotko"]
})
print (df.dtypes)
```

```python
df["level"] = df["level"].astype('category')
```

```python
df['level'] = df['level'].cat.rename_categories({
    "junior": "jr"
})
```

```python
df['level_code'] = df['level'].cat.codes
print(df.head())
```

```python
#this is one hot encoding
encoded = pd.get_dummies(df['level'], prefix='lev')
df = pd.concat(obs= [df, encoded], axis=1)
print(df.head())

```

```python
df['level_code'] = df['level'].cat.codes
df = pd.concat(df, encoded)
```

# handling time series data

```python
import pandas as pd
df = pd.DataFrame({
    'Date': ['2025-02-2', '2025-03-23','2025-12-5','2025-07-7'],
    'Sales': [100, 1221, 12,324]

})
print(df.dtypes)
```

```python
df['Date'] = pd.to_datetime(df['Date'])
print(df.dtypes)
```

```python
df = df.set_index('Date')
print(df.head())
```

```python
print(df.head())
```

```python
df['year'] = df.index.year
df['month'] = df.index.month
df['day'] = df.index.day
df['day_name'] = df.index.day_name()
df['month_name'] = df.index.month_name()
print(df)
```
