# feature encoding
- nominal -> unordered categorical data. eg cities, colors etc
- ordinal -> ordered categorical data. eg cloth_size

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
cities = ['Ktm', "dharan", "pokhara"]
encoded = le.fit_transform(cities)
print(encoded)
```

```python
# one hot encoding
import pandas as pd

df = pd.DataFrame({'source': ['river', 'lake', 'well']})
df_encoded = pd.get_dummies(df, columns=['source'])
print(df_encoded)
```

```python
#ordinal encoding

from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(categories=[['poor', 'fair','good']])
quality = ['poor', 'fair','good']
encoded = encoder.fit_transform(quality)
print(encoded)
```
