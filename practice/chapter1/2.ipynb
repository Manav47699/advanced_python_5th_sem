# Advanced data structures: Collections, Iterators, Generators, and Decorators



Collections: It is a built-in python module that provides us with more powerful structures thann traditional lists and dicts.

- *Counter*: It is used to count how many times an item appeared in a iterable(eg: list, tuple, string, generator). 

When we pass a iterable through Counter() it returns a Counter object (a subclass of dicts) with the elements as keys and their frequencies as the values.
 

```python
from collections import Counter

fruits = ["apple", "banana", "apple", "grapes"]
c = Counter(fruits)
print (c)
```

- defaultdict is just like a normal dictionary, but it automatically creates a default value for a missing key.

First, we create its object and pass a data type (actually a callable function) as an argument.
When a key that is not already in the dictionary is accessed, it is automatically added to the dictionary with a default value.

```python
from collections import defaultdict

d = defaultdict(int)
d['a'] += 1   # d['a'] does not exist, so it will be created at d['a']:0. then we did +1 which will give us the output
print (d)
```

- deque: Double Ended queue. This is very usefull for adding or removing elements from a queue from both sides

instead of appendright(), we have append() 

instead of popright(), we have pop()

```python
from collections import deque

dq = deque([1, 2, 3, 4, 5])
dq.appendleft(0)
dq.append(3)
print (dq)
dq.pop()
dq.popleft()
print(dq)
```

- namedtuple: it is a subclass of tuple, where we can also use name (instead of just index) to acess the elements making it like a mixture of tuple and object.

it works by making a class dynamically. like in the below example, a class named 'Point' is created automatially.

in the below example, a class named "Car" is created where "name", "company" and "age" are attributes and "c" is the object.



```python
from collections import namedtuple

Car = namedtuple('Car', ["name", "company", "model"])
c = Car("ecosport", "ford", 2025)

print (c.name, c.model, c.company)
```

# Iterators [iter()]
-> An iterator is an object created by passing an iterable structure to  iter(), which returns elements one by one using next(). When no elements remain, next() raises StopIteration.

```python
numbers = [1, 2, 3, 4, 6, 7]
it = iter(numbers)

while True:
    try:
        x = next(it)
        print (x)
    except StopIteration:
        break
```

# Genarators
-> Generator is a simple way to build an iterator with "yield". It saves memory as it produces values only when needed.

```python
def square(n):
    for i in range (n):
        yield i * i

for nums in square(5):
    print (nums)
```

```python
def add(a,b):
    yield a+b

print (add(4, 6))
```

# Decorators
-> It is used to add an extra feature to any function without actually changing the function block.



```python
def my_dec(func):
    def wrapper():
        print ("before")
        func()
        print ("after")
    return wrapper

@my_dec
def hellow():
    print ("heloowwww")

hellow()
```
