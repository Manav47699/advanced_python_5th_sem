# functions and lambda expressions

# Lambda function
-> It is a small anonymous function used when the function is short and simple.

- Syntax: "function_name = lambda arg1,arg2 : code

```python
sum = lambda a, b : a+b
print (sum(3, 2))
```

# common uses of lambda function

- Filtering: Eg: filtering odd and even from a list of numbers

- Syntax: list(filter(lambda_function, iterable))

Here the lamda function is generally a conditinal statement which will return either Ture of False. It will return only the elements from the iterable for which the function returns true

```python
nums = [1, 2, 3, 4,5 ,6, 7, 8, 9]
evens = list(filter(lambda x: x%2 == 0, nums))
print (evens)
```

- mapping: this will go through each element in an iterable, apply the lamda function and then return the updated iterable.

- Syntax: list(map(lambda_function, iterable)) 

```python
nums = [1, 2, 3, 4,5 ,6, 7, 8, 9]
squares = list(map(lambda x: x*x, nums))
print (squares)
```
