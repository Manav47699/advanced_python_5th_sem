# Exception handling, debugging, and logging

# Exception Handeling
-> When a program runs into an error, execption handeling can be used to handle then gracefully

```python
# Catching multiple/ all exceptions
try:
    num = int(input(" enter the denominatior"))
    print (f"the number you entered was {num}")
    num = 10/0
except ValueError:
    print ("Please enter a valid number")
except ZeroDivisionError:
    print ("division by zero is not allowd")
except Exception as e:
    print ("some random exception has occured.")
```

```python
# try-expect-else-finally block

try:
    x = 10/2
except ZeroDivisionError:
    print ("cant divide by zero")
else:
    print (f"no errors, Answer= {x}")
finally:
    print("this code will run no matter what.")
```

# Debugging:
We can use the pdb module (Python Debugger) to debug our code.

By calling pdb.set_trace() (or breakpoint() in modern Python) inside a program, execution pauses at that line and enters interactive debug mode in the terminal.

In debug mode, we can use commands like:

n → move to next line

s → step into a function

c → continue execution

p variable → print variable value

q → quit debugger

```python
import pdb

def divide(a, b):
    pdb.set_trace()
    return a / b

divide(10, 1)
```

# Logging
-> Logging records program events and errors using different severity levels instead of print statements.

- there are levels to logging (debug, info, warning, error, critical)

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("Debug message")
logging.info("Information message")
logging.warning("Warning message")
logging.error("Error message")
logging.critical("Critical message")
```

- the below example will append the output to a file name app.log


```python
import logging

logging.basicConfig(
    level=logging.DEBUG, 
    format="%(asctime)s %(levelname)s %(message)s", 
    filename="app.log",
    filemode="a"
)

logging.debug("Debug message")
logging.info("Information message")
logging.warning("Warning message")
logging.error("Error message")
logging.critical("Critical message")


# output (yo chai ipynb ma chaldaina tei vara yei output lekhdeko haita)
'''
2026-02-19 16:25:00,123 DEBUG Debug message
2026-02-19 16:25:00,124 INFO Information message
2026-02-19 16:25:00,125 WARNING Warning message
2026-02-19 16:25:00,126 ERROR Error message
2026-02-19 16:25:00,127 CRITICAL Critical message
'''
```
