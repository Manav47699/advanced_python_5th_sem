#  Web scraping using requests and BeautifulSoup

# Using requests to send and HTTP request and getting the html

```python
import requests

url = "https://www.scrapethissite.com/page/simple"
response = requests.get(url)
print (response.status_code)
print(response.text)
```

# Using bs4 to parse the html
-> Parsing means taking the raw and messy file and make a structured format like a tree.

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(response.text, "html.parser")
```

# Finding elements by tags

```python
titles = soup.find_all("h1")
for t in titles:
    print (t.text)
```

# Finding elements by classes

```python
items = soup.find_all("p", class_="content")  # "class" hoina "class_" ho haita
for c in items:
    print (c.text)
```

```python
titles = soup.find_all("a")
for t in titles:
    print (t.get("href"))
```

# Saving the scraped data

```python
import csv

with open("links.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(['title', 'link'])
    for t in titles:
        writer.writerow([t.text.strip(), t.get("href")])

```
