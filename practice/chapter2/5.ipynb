# Handling large datasets with chunking and lazy evaluation


# Lazy Evaluation

```python
def read_lines(filename):
    with open(filename) as f:
        for line in f:
            yield line

for line in read_line("data.csv"):
    print (line)
```

# Chunking

```python
import pandas as pd
import os

for chunk in pd.read_csv("data.txt", chucksize=4):
    print(chunk.head())
```
