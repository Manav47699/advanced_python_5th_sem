# Merging 
-> It combines DataFrames iusing a common column
- inner: If a key exits in both tables, it stays.
- left: keeps everything from the left table, matches it with the right table and keeps (NaN) if nothing exists 
- right: keeps everything from the right table, matches it with the right table and keeps (NaN) if nothing exists 
- outer: keeps everything from both the tabels and keeps (NaN) if nothing exists.

```python
import pandas as pd

df_students = pd.DataFrame({
    'Student_ID': [1, 2, 3],
    'Name': ["Manav", "Mausam", "himal"]
})

df_marks = pd.DataFrame({
    'marks': [9, 8, 10],
    'Student_ID': [2, 3, 4]
})

print (df_students)
print (df_marks)


#inner merge
merged_inner = pd.merge(df_students, df_marks, on="Student_ID", how="inner")
print("Inner merge is given below:")
print(merged_inner)

# left merge
merged_left = pd.merge(df_students, df_marks, on="Student_ID", how="left")
print("Left merge is given below:")
print(merged_left)

# right merge
merged_right = pd.merge(df_students, df_marks, on="Student_ID", how="right")
print("Right merge is given below:")
print(merged_right)

# outer merge
merged_outer = pd.merge(df_students, df_marks, on="Student_ID", how="outer")
print("Outer merge is given below:")
print(merged_outer)
```
