# Merging 
-> It combines DataFrames iusing a common column
- inner: If a key exits in both tables, it stays.
- left: keeps everything from the left table, matches it with the right table and keeps (NaN) if nothing exists 
- right: keeps everything from the right table, matches it with the right table and keeps (NaN) if nothing exists 
- outer: keeps everything from both the tabels and keeps (NaN) if nothing exists.

```python
import pandas as pd

df_students = pd.DataFrame({
    'Student_ID': [1, 2, 3],
    'Name': ["Manav", "Mausam", "himal"]
})

df_marks = pd.DataFrame({
    'marks': [9, 8, 10],
    'Student_ID': [2, 3, 4]
})

print (df_students)
print (df_marks)


#inner merge
merged_inner = pd.merge(df_students, df_marks, on="Student_ID", how="inner")
print("Inner merge is given below:")
print(merged_inner)

# left merge
merged_left = pd.merge(df_students, df_marks, on="Student_ID", how="left")
print("Left merge is given below:")
print(merged_left)

# right merge
merged_right = pd.merge(df_students, df_marks, on="Student_ID", how="right")
print("Right merge is given below:")
print(merged_right)

# outer merge
merged_outer = pd.merge(df_students, df_marks, on="Student_ID", how="outer")
print("Outer merge is given below:")
print(merged_outer)
```

# Joining
-> It combines DataFrames using index instead of columns like Merging.

- Inner join: keeps only rows with matching index in both DataFrames

- Left join: keeps all rows from left DataFrame, fills unmatched with NaN

- Right join: keeps all rows from right DataFrame, fills unmatched with NaN



```python
df_A = pd.DataFrame({'Temperature': [20, 40, 45]},
                    index=['Day1', 'Day2', "Day3"])
df_B = pd.DataFrame({'Humidity': [60, 65, 70]},
                    index=['Day1', 'Day2', "Day4"])

#inner join
joined_inner = df_A.join(df_B, how="inner")
print (joined_inner)

#left join
joined_left = df_A.join(df_B, how="left")
print (joined_left)

#right join
joined_right = df_A.join(df_B, how="right")
print (joined_right)
```

# Reshaping
-> It means changing the structure of a Dataframe keeping the values same.

- melt() → Converts wide format to long format (columns → rows).

- pivot() → Converts long format to wide format (rows → columns) without aggregation.

- pivot_table() → Same as pivot, but allows aggregation (mean, sum, count, etc.) when duplicates exist.

- stack() → Moves column levels into the row index (columns → inner row index).

- unstack() → Moves row index levels into columns (index → columns).

# Wide VS Long data format

# Wide data format
-> Each variable has a separate column.

# Long data format
-> Variables are stored in a row and not a seperate column.

```python
# this is how wide format is written
df_wide = pd.DataFrame({
    'Day': ['Mon', 'Tue'],
    'Temp': [20, 22],
    'Humidity': [60, 65]
})
print (df_wide)

# this is how long format is written
print ("\n  \n")
df_long = pd.DataFrame({
    'Day': ["Mon", "mon","tue", "tue"],
    'varibale': ["temp", "humidity", "temp", "Humidity"],
    'Value': [20, 30 , 40,50]
})
print (df_long)

```

# Melt 

- id_vars: columns to keep fixed
- var_name: name of new variable column
- value_name: name of value column


```python
df_wide = pd.DataFrame({
    'Day': ['Mon', 'Tue'],
    'Temp': [20, 22],
    'Humidity': [60, 65]
})

df_melt = pd.melt(df_wide, id_vars=['Day'], var_name='Weather', value_name='Score')
print (df_melt)
```

# Pivot

```python
df_long = pd.DataFrame({
    'Day': ["Mon", "mon","tue", "tue"],
    'varibale': ["temp", "humidity", "temp", "Humidity"],
    'Value': [20, 30 , 40,50]
})

df_pivot= pd.pivot(df_long, index='Day', columns='varibale', values='Value')
print(df_pivot)
```

# Stack


```python
df_long = pd.DataFrame({
    'Day': ["Mon", "mon","tue", "tue"],
    'varibale': ["temp", "humidity", "temp", "Humidity"],
    'Value': [20, 30 , 40,50]
})

df_stacked = df_long.stack()
print(df_stacked)
```

# Pivot Table

```python
df_long = pd.DataFrame({
    'Day': ["Mon", "mon","tue", "tue"],
    'varibale': ["temp", "humidity", "temp", "Humidity"],
    'Value': [20, 30 , 40,50]
})

df_pivot_table = pd.pivot_table(df_long, 
                                index='Day', 
                                columns='varibale', 
                                values='Value', 
                                aggfunc='mean')

print (df_pivot_table)
```
